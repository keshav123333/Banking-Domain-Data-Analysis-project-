```python
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
import numpy as np
```

```python
x=np.random.uniform(1,10,size=(100,1))
y=np.random.randint(0,2,size=(100,))
```

```python
x.shape,y.shape
```

```python
xtest=x[:50,:]
ytest=y[:50]
x=x[50:,:]
y=y[50:]
```

```python
x.shape,y.shape
```

```python
model1=LogisticRegression()
model1.fit(x,y)
```

```python
pred=model1.predict(xtest)
```

```python
accuracy_score(ytest,pred)
```

#Confusion matrix

```python
w=confusion_matrix(ytest,pred)
```

```python
import seaborn as sns
sns.heatmap(w,annot=True)
```

type 1 error-> mean false postitve

type2 error -> false neg

#Multiclass classifcation

isme precison weighted and macro hota hai aur we can say har class ka apna precison recall

```python
x=np.random.uniform(1,10,size=(500,1))
y=np.random.randint(0,3,size=(500,))
```

```python
xtest=x[:250,:]
ytest=y[:250]
x=x[250:,:]
y=y[250:]
```

```python
model2=LogisticRegression()
model2.fit(x,y)
```

```python
pred1=model2.predict(xtest)
```

```python
sns.heatmap(confusion_matrix(ytest,pred1),annot=True)
```

```python
precision_score(ytest,pred1,average='weighted'),recall_score(ytest,pred1,average='weighted')
```

```python
f1_score(ytest,pred1,average="weighted")
```

```python
from sklearn.metrics  import classification_report
print(classification_report(ytest,pred1))
```
